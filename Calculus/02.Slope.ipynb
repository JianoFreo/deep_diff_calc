# Slope of line

# $\frac{rise}{run}$ = $\frac{\Delta y}{\Delta x}$

### this means every time we move forward on the x axis by 1 unit, we move up on the y axis by only half of that 

### one step forward on the x axis is 1 unit, and one step up on the y axis is 0.5 units

### one step upward is two step forward mean 1/2 slope

### because every 2 step upward on the y axis is 1 step forward on the x axis, we can say that the slope of this line is 2
