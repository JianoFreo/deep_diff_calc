# Changes to the data

if you are going to change the data make sure you use an assignement operator to dataframe to make sure the changes are saved. Otherwise, the changes will be temporary and will not be reflected in the original data frame.

well not in all cases like when you are using inplace = True, but in general, it is a good practice to use an assignment operator to make sure the changes are saved.


also this one doesnt need an assignment variable = df.loc[0:5, 'Name'] = "Ivysaur"

===================================================================
## “Functions usually DO NOT change df unless told”

If you see a function like:

df.drop()

df.sort_values()

df.fillna()

df.rename()

It returns a NEW DataFrame

## but if you want it to modify the dataframe directly then you need to use inplace = True

### Option A: reassign

df = df.drop(columns=['Total'])

### Option B: inplace (rarely recommended)

df.drop(columns=['Total'], inplace=True)

=================================================================
## “Assignment ALWAYS changes the DataFrame”

If you see = with .loc / .iloc:

df.loc[0:5, 'Name'] = "Ivysaur"

df.iloc[0, 1] = 10

df['col'] = something

This modifies df directly


## add a column

```python
df['Total' ] = df['HP' ] + df['Attack'] + df['Defense'] + df['Sp. Atk'] + df['Sp. Def' ] + df['Speed']
```

```python
df.head(5)
```

## drop a column

```python
df = df.drop(columns=['Total'])
```

```python
df.head()
```

```python
df['Total'] = df.iloc[:,4:10].sum(axis=1) 
#[:] means all rows
#[4:9] means “Take columns 4 to 9 and sum them row by row, store in Total”

# iloc indexing starts  at 0 thats why we start at 4 instead of 5
# 0 → #
# 1 → Name
# 2 → Type 1
# 3 → Type 2
# 4 → HP
# 5 → Attack
# 6 → Defense
# 7 → Sp. Atk
# 8 → Sp. Def
# 9 → Speed
# 10 → Generation
# 11 → Legendary
# 12 → Total
df
```

# moving the columns around

in this case we are moving the total

```python
cols = list(df.columns)

df = df[cols[0:4] + [cols[-1]] + cols[4:12]]

df.head(5)

##====== my mistake ==========
# df = df[cols[0:4] + cols[-1] + cols[4:12]]

# This part is wrong:

# cols[-1]

# Because:

# cols[-1] is a single string (e.g. "Total")
# you cannot add a string to a list

# So this will crash or behave incorrectly.

## cols[0:4] and cols[4:12] are not strings because they are a range therefore a list
```
