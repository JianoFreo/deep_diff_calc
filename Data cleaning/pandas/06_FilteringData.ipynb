# with and without loc

Your two lines

## A) Boolean filtering (most common)

df[df['Type 1'] == 'Grass'].head(5)

## B) .loc version

df.loc[df['Type 1'] == 'Grass'].head(5)

more explicit pandas method

label-based indexer

allows row + column selection together



## Key answer first

    In this case, both return the same result

    BUT the way they work internally is different.

| Feature                   | `df[...]` | `df.loc[...]`  |
| ------------------------- | --------- | -------------- |
| readability               | simpler   | clearer intent |
| column selection          | limited   | powerful       |
| row + column filtering    | no        | yes            |
| recommended in interviews | yes       | YES (better)   |


# Important insight
This works:
```python
df[df['Type 1'] == 'Grass']
```
But this is more powerful:
```python
df.loc[df['Type 1'] == 'Grass', ['Name', 'HP']]
```
## .loc = full control

Why .loc is preferred in serious code

Because it avoids ambiguity:

df[...] → could be row or column confusion

df.loc → explicitly row + column selector

======================================================


## One important subtle difference

If you later do assignment:
```python
df.loc[df['Type 1'] == 'Grass', 'HP'] = 0
```
It sets the HP of all Grass-type Pokémon to 0

✔ modifies safely

But:
```python
df[df['Type 1'] == 'Grass']['HP'] = 0
```
may fail (chained indexing problem)

# Final rule

.loc is safer and more powerful

df[...] is shorter but limited

===========================================================
# To change multiple columns in pandas, you just pass a list of column names instead of a single column.

## Basic idea
```python
df.loc[rows, [col1, col2, col3]] = value
```
Example 1: change multiple columns to same value

```python
df.loc[df['Type 1'] == 'Grass', ['HP', 'Attack']] = 0
```
sets HP and Attack to 0 for Grass Pokémon

Here are some pandas filtering challenges using a Pokémon dataset (assuming columns like `Name`, `Type 1`, `Type 2`, `HP`, `Attack`, `Defense`, `Speed`, `Legendary`, `Generation`).

Try writing the `.loc[]` filters yourself before checking the answers.

### Easy

1. Get all Grass Pokémon.
2. Get all Pokémon with HP greater than 100.
3. Get all Legendary Pokémon.
4. Get all Pokémon from Generation 1.
5. Get only the `Name` and `Attack` columns for Fire Pokémon.

```python
# Example format
df.loc[df['Type 1'] == 'Grass']
```



````python
df.loc[(df['Type 1'] == 'Grass') | (df['Type 2'] == 'Grass')].sample(10).sort_values('Name' , ascending=True)
````

````python
df.loc[df['HP'] > 100].sample(4)
````

````python
df.loc[~df['Legendary']].sample(5)
````

````python
df.loc[df['Legendary']].sample(5)
````

````python
df.loc[df['Generation'] == 1].sample(4)
````

````python
df.loc[(df['Type 1'] == 'Fire') | (df['Type 2'] == 'Fire'),['Name', 'Attack']]
````

# Medium
### Get all Water Pokémon with HP above 80.
### Get all Pokémon whose Attack is greater than Defense.
### Get all Pokémon that are either Fire or Electric.
### Get all non-Legendary Pokémon.
### Get Pokémon with Speed between 80 and 120.

````python
df.loc[(df['HP'] > 80) & ((df['Type 1'] == 'Water') | (df['Type 2'] == 'Water'))].sample(10).sort_values('Name', ascending=True)
````

````python
df.loc[df['Attack'] > df['Defense']].sample(5)
````

````python
df.loc[
    (df['Type 1'].isin(['Fire', 'Electric'])) |
    (df['Type 2'].isin(['Fire', 'Electric']))
].sample(10).sort_values('Name')
````

````python
df.loc[(df['Speed'] > 80) & (df['Speed'] < 120)]
# ==================OR========================
df.loc[df['Speed'].between(80, 120)].sample(10)
````
