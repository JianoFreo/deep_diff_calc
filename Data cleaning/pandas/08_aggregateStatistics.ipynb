# Aggregate Statistics

```python
import pandas as pd
```

```python
df = pd.read_csv('../data/pokemon_data.csv')
df
```

# group by

```python
df.groupby(['Type 1']).mean(numeric_only=True).round()
```

```python
#combine the type 1 and type 2
g1 = df.groupby('Type 1').mean(numeric_only=True)
g2 = df.groupby('Type 2').mean(numeric_only=True)

result = g1.add(g2, fill_value=0).round(2)
result
```

```python
df.groupby(['Type 1', 'Type 2']) \
    .mean(numeric_only=True) \
    .sort_index(level=['Type 1', 'Type 2'], ascending=[True, False]).round(2) 
    # sort_index because the types are indexes nnow beacuse wee grouped them by it
```

```python
df.loc[(df['Type 1'] == 'Bug') & (df['Type 2'] == 'Electric')].mean(numeric_only=True)
```

```python
df.loc[df['Attack'].astype(str).str.contains(r'12', na=False)].sample(5)
```

```python
df['count'] = 1
df.groupby(['Type 1']).count()['count']
```

```python
# wwould also add the type 2

df2 = df.copy()

df2['Type'] = df['Type 1']
df3 = df.copy()
df3['Type'] = df['Type 2']

combined = pd.concat([df2, df3])

combined.groupby('Type').size()
```
