```python
-- Active: 1780564061414@@127.0.0.1@3306@pokemon
-- Active: 1780564061414@@127.0.0.1@3306@mysql
CREATE TABLE pokemon (
    id INT,
    Name VARCHAR(100),
    Type1 VARCHAR(50),
    Type2 VARCHAR(50),
    HP INT,
    Attack INT,
    Defense INT,
    SpAtk INT,
    SpDef INT,
    Speed INT,
    Generation INT,
    Legendary BOOLEAN
);
```

```python
LOAD DATA LOCAL INFILE 'C:/Users/User/Desktop/deep_diff_calc/Data cleaning/data/pokemon_data.csv'
INTO TABLE pokemon
FIELDS TERMINATED BY ','
IGNORE 1 ROWS
```

```python
SELECT Name, Type1
FROM pokemon
WHERE Name = 'Ivysaur';
```

```python
SELECT Name, Type1, Type2
FROM pokemon
WHERE Type2 IS NULL
or Type2 = ''
or TRIM(Type2) = ''
order by Name
```

```python
SELECT Name, Type1, Type2
FROM pokemon
WHERE Type2 IS NULL
or Type2 = ''
or TRIM(Type2) = ''
order by Name
Limit 5
```

```python
Select Name, Type2
from pokemon
order By name asc
```

```python
SELECT Name,
       Type1,
       CONCAT('No type 2 but I have ', Type1) AS Type2
FROM pokemon
WHERE Type2 IS NOT NULL
  AND TRIM(Type2) != ''
ORDER BY Name;
```
